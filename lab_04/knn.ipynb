```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, classification_report
import matplotlib.pyplot as plt
from sklearn.datasets import make_classification
```

```python
X, y = make_classification(n_samples=200, n_features=2, n_informative=2, n_redundant=0,
                           n_clusters_per_class=1, random_state=42, class_sep=1.5)

df = pd.DataFrame(X, columns=['feature_1', 'feature_2'])
df['target'] = y

print("First 5 rows of the dataset:")
display(df.head())

plt.figure(figsize=(8, 6))
plt.scatter(X[:, 0], X[:, 1], c=y, cmap='viridis', s=50, edgecolors='k')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.title('Synthetic Classification Data')
plt.colorbar(label='Class')
plt.show()
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

print(f"Training set size: {X_train.shape[0]} samples")
print(f"Testing set size: {X_test.shape[0]} samples")
```

```python
k = 5
knn = KNeighborsClassifier(n_neighbors=k)

knn.fit(X_train, y_train)

print(f"KNN classifier trained with k={k} neighbors.")
```

```python
y_pred = knn.predict(X_test)

print("First 10 predictions on the test set:")
print(y_pred[:10])
print("First 10 actual values on the test set:")
print(y_test[:10])
```

```python
accuracy = accuracy_score(y_test, y_pred)
report = classification_report(y_test, y_pred)

print(f"Accuracy: {accuracy:.2f}")
print("\nClassification Report:")
print(report)
```
